```python
# Importing Libraries
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
def salaty_type(salary):
    if salary < 60_000:
        return 'Low'
    elif 60_000 <= salary <= 100_000:
        return 'Medium'
    else:
        return 'High'
        

df_filtered = df.copy()
df_filtered.dropna(subset= 'salary_year_avg', inplace = True)

df_filtered['salary_category'] = df_filtered['salary_year_avg'].apply(salaty_type)

df_filtered[['salary_year_avg', 'salary_category']].head()
```
