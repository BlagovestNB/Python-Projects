```python
# Importing Libraries
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
oct_df = df[df['job_posted_date'].dt.strftime('%B') == 'October']

nov_df = df[df['job_posted_date'].dt.strftime('%B') == 'November']

dec_df = df[df['job_posted_date'].dt.strftime('%B') == 'December']

combined_df = pd.concat([oct_df, nov_df, dec_df], ignore_index= True)

combined_df.sample(20)



```
